# Drawing QA - Start Here

This notebook checks a drawing package for you. It **never changes your drawings**, it only reports problems.

**How to use it**
- Click a grey box (a "cell") and press **Shift + Enter** to run it.
- Run the boxes **from top to bottom**.
- When a box is running you will see `[*]` next to it. Wait until it shows a number.

## Step 1 - Install (first time only)

Run this once on a new computer. It takes a few minutes.

When it finishes, **restart the notebook**: click **Restart** in the toolbar at the top (or *Kernel → Restart*), then carry on with Step 2.

```python
%pip install -r requirements.txt
```

## Step 2 - Choose the package to check

Put your drawings in a folder inside `input`, with these sub-folders (any can be missing):

`PLC` · `CableBlocks` · `Terminations` · `EquipmentList`

Then type the folder name between the quotes below and run the box.

```python
PACKAGE = "input/Main"

from drawing_qa.cli import main
print("Ready to check:", PACKAGE)
```

## Step 3 - Survey (quick look)

Counts the sheets and tells you how many can be read. It does not check anything. Takes a few seconds.

```python
main(["survey", "--input", PACKAGE]);
```

## Step 4 - Run the check

Reads every sheet and looks for problems. This can take a minute or two.

```python
main(["run", "--input", PACKAGE]);
```

## Step 5 - Open the results

Opens the results folder. The files you will use most:
- **QA_Register.xlsx** - the list of problems, to print and mark up
- **report.md** - the summary
- **system_map.html** - a map of the system, opens in your web browser

```python
import os
from pathlib import Path

results = Path("reports") / Path(PACKAGE).name
os.startfile(results)
```
